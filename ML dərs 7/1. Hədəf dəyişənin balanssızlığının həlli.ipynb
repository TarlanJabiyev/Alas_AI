```python
import pandas as pd
from imblearn.under_sampling import NearMiss
from imblearn.over_sampling import SMOTE
from imblearn.over_sampling import BorderlineSMOTE
from imblearn.over_sampling import SVMSMOTE
```

```python
data = pd.read_csv('churn.csv')
```

```python
data
```

```python
data.info()
```

```python
data['y'].value_counts()
```

```python
data['y'].value_counts(normalize=True)
```

```python
no = len(data[data['y'] == 'no'])
yes = len(data[data['y'] == 'yes'])
yes/no
```

```python
x = data.select_dtypes(include=['number']).drop(columns=['id'])  
y = data['y']  
```

## Near Miss Under Sampling

```python
undersampling = NearMiss(version=1, n_neighbors=3)
undersampling
```

```python
X, Y = undersampling.fit_resample(x, y)

Y.value_counts()
```

```python
undersampling = NearMiss(version=2, n_neighbors=3)
undersampling
```

```python
X, Y = undersampling.fit_resample(x, y)

Y.value_counts()
```

```python
undersampling = NearMiss(version=3, n_neighbors_ver3=3)
undersampling
```

```python
X, Y = undersampling.fit_resample(x, y)

Y.value_counts()
```

## SMOTE (Synthetic Minority Over Sampling Technique)

```python
oversample = SMOTE()
oversample
```

```python
X, Y = oversample.fit_resample(x, y)

Y.value_counts()
```

## Borderline-SMOTE

```python
oversample = BorderlineSMOTE()
oversample
```

```python
X, Y = oversample.fit_resample(x, y)

Y.value_counts()
```

## Borderline-SMOTE SVM

```python
oversample = SVMSMOTE()
oversample
```

```python
X, Y = oversample.fit_resample(x, y)

Y.value_counts()
```
